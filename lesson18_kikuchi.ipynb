```python
# 選択法
def selection_sort(li):
    for i in range(len(li) - 1):
        for j in range(i + 1, len(li)):
            if li[i] > li[j]:
                tmp = li[i]
                li[i] = li[j]
                li[j] = tmp

    return li
```

```python
# 交換法
def bubble_sort(li):
    for i in range(len(li) - 1):
        for j in range(len(li) - 1, i, -1):
            if li[j - 1] > li[j]:
                tmp = li[j - 1]
                li[j - 1] = li[j]
                li[j] = tmp
```

```python
# 挿入法
def insertion_sort(li):
    new_li = []

    for i in range(len(li)):
        new_li.append(li[i])

        for j in range(i, 0, -1):
            if new_li[j - 1] > new_li[j]:
                tmp = new_li[j - 1]
                new_li[j - 1] = new_li[j]
                new_li[j] = tmp
            else:
                break

    return new_li
```

```python
li = [4, 2, 3, 5, 1]
li = insertion_sort(li)  # 実行したいソートの関数名に書き換えること
print(li)
```

日経平均株価のページのHTMLデータを取得してBeautifulSoupのオブジェクトを生成する

```python
import urllib.request as request
from bs4 import BeautifulSoup
import pandas as pd
from datetime import datetime, timedelta, timezone

url = 'https://www.nikkei.com/markets/worldidx/chart/nk225/'

response = request.urlopen(url)

bs = BeautifulSoup(response, 'html.parser')
```

```python
# 株価を取得
stocksPrice = bs.select('.economic_value_now')[0].text
print(stocksPrice)
```

```python
# 今日の日付を取得
jst = timezone(timedelta(hours=+9), 'JST')
today = datetime.now(jst).date().isoformat()

# PandasのDataFrameを生成
df = pd.DataFrame(data = [[today, stocksPrice]])

# CSVに保存（追記）
df.to_csv('stoks_price.csv', mode = 'a', header = False, index = False)
```
