```python
def factorial(n):
    if n < 2:
        return 1
    
    return n * factorial(n - 1)
   
```

```python
factorial(0)
```

```python
def fib(i):
    if i == 0:
        return 0
    if i == 1:
        return 1
    if i == 2:
        return 1    
    
    return fib(i-1) + fib(i-2)
```

```python
fib(6)
```

```python
def fibo(i):
    if i == 0:
        return 0
    if i == 1:
        return 1
    if i == 2:
        return 1    
    
    return fibo(i-1) + fibo(i-2)
```
