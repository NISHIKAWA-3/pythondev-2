```python
class Dog:
    # property
    voice = "bow!"

    # constructor
    def __init__(self, name = "Buddy"):    #コンストラクタ __init__ のなかでメンバ変数 name を定義.引数にある name = "Buddy" が、その定義部分
        # property
        self.name = name  #指定された名前をメンバ変数 name へ代入するよう初期処理として定義.  
                          #左辺の self.name が Dog クラスのインスタンスメンバ変数 name、右辺の name はコンストラクタの引数に指定された変数 name
                          #インスタンスメンバ変数となる

    # method
    def bark(self):
        print(self.voice)
```

```python

buddy = Dog()  #Dogクラスのインスタンスを生成
print(buddy.name)  #インスタンス変数名.name で名前を参照
buddy.bark()  #インスタンス変数名.bark() で bark() メソッドを実行

jake = Dog("Jake") #Dogクラスのインスタンスを生成
print(jake.name) #インスタンス変数名.name で名前を参照
jake.bark() #インスタンス変数名.bark() で bark() メソッドを実行
```

```python
class Suit:
    def __init__(self, color, size, brand):
        self.color = color
        self.size = size
        self.brand = brand

    def __call__(self, occasion):
        return f"この{self.color}のスーツ（サイズ：{self.size}、ブランド：{self.brand}）は、{occasion}にぴったりです。"

# クラスのインスタンス化
my_suit = Suit(color="ピンク", size="L", brand="サンプルブランド")

# インスタンスを関数のように呼び出す
result = my_suit("パーティー")
result
```

```python
# インスタンスを関数のように呼び出す
result = my_suit.__call__("a formal event")
result
```

```python
class Dog:
    # property
    voice = "bow!"

    # constructor
    def __init__(self, name = "Buddy"):
        self.name = name

    # method
    def bark(self):
        print(self.voice)

    # class method
    @classmethod
    def description(self):
        print(f"犬の鳴き声は{self.voice}")
```

```python
class Dog:
    # property
    voice = "bow!"  #メンバ変数（プロパティ）（クラスメンバ変数)

    # constructor
    def __init__(self, name = "Buddy"): #__int__の中でメンバ変数nameを定義（name="Buddy")
        self.name = name #指定された名前をメンバ変数nameへ代入するよう定義。インスタンスメンバ変数。

    # method
    def bark(self):  #methodの定義、一番目の引数にselfを入れる.インスタンスメソッドという扱い
        print(self.voice) #self.メソッド名（）で自身が持つメンバ変数やメソッドを実行

    # class method
    @classmethod 
    def description(self): #Dog クラスに「犬の鳴き声はbow!」と説明する description() メソッドをクラスメソッドとして定義
        print(f"犬の鳴き声は{self.voice}")

class Japanese_Dog:
    voice = "ワン！"

    def __init__(self, name = "ポチ"):
        self.name = name

    def bark(self):
        print(self.voice)

    @classmethod
    def description(self):
        print(f"日本の犬の鳴き声は{self.voice}")
```

```python
Dog.description() # この行ではインスタンスを生成していない

buddy = Dog() #Dogクラスのインスタンスを生成
print(buddy.name) #インスタンス変数名.name で名前を参照
buddy.bark() #インスタンス変数名.bark() で bark() メソッドを実行

jake = Dog("Jake")
print(jake.name)
jake.bark()

Japanese_Dog.description() # この行ではインスタンスを生成していない

pochi = Japanese_Dog()
print(pochi.name)
pochi.bark()

shiro = Japanese_Dog("シロ")
print(shiro.name)
shiro.bark()
```

```python
#親クラス
class Dog: 
    def __init__(self, name): #インスタンスメンバ変数nameを定義
        self.name = name

    def bark(self):  #bark() を定義
        print(self.voice)

#子クラス
class English_Dog(Dog): #継承の形をとる場合は、括弧を追加して親クラス名を中に書く形（書式：class クラス名(親クラス名):）とする
    voice = "bow!"

    def __init__(self, name = "Buddy"):
        super().__init__(name) #super().メソッド名() と書くことで、親クラスのメソッドを実行
                               #親クラスである Dog のコンストラクタを name の値だけ渡して実行する

    @classmethod
    def description(self):
        print(f"英語圏の犬の鳴き声は{self.voice}")

class Japanese_Dog(Dog):
    voice = "ワン！"

    def __init__(self, name = "ポチ"):
        super().__init__(name)

    @classmethod
    def description(self):
        print(f"日本の犬の鳴き声は{self.voice}")
        
```

```python
English_Dog.description()

buddy = English_Dog()
print(buddy.name)
buddy.bark()

jake = English_Dog("Jake")
print(jake.name)
jake.bark()

Japanese_Dog.description()

pochi = Japanese_Dog()
print(pochi.name)
pochi.bark()

shiro = Japanese_Dog("シロ")
print(shiro.name)
shiro.bark()
```

```python
tanaka = Human2("田中")
print(tanaka.__name) #__name として変数を定義したはずなのに、その変数が見つからないというエラーが出るのは、カプセル化の影響
```

```python
tanaka = Human2("田中")
print(tanaka._Human2__name) #インスタンス変数名._クラス名__メンバ変数名で入力すると実行可能。でもオフィシャル仕様ではない。
```

```python
#オフィシャル仕様ゲッター および セッター と呼ばれるメソッド経由
class Human2:
    def __init__(self, name):
        self.set_name(name)

    # ゲッター
    def get_name(self):
        return self.__name

    # セッター
    def set_name(self, name):
        self.__name = name
```

```python
tanaka = Human2("田中")

# ゲッターを使用して値を取得
print(tanaka.get_name())
```
