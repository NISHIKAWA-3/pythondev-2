```python
tpl=(1,2)+(3,4)#リストと同じく、+ 演算子を用いてタプル同士の結合を行うことができます。
print(tpl)
print(len(tpl))#リストと同じく len() 関数でタプルに含まれている要素の数を取得することができます
#タプルとリストの違いとして、タプルは一度作成したオブジェクトに編集を加えることができません。
#例えば、タプルには append メソッドや remove メソッドがないため、次のコードはエラーが発生
```

```python
st={1, 2, 3}
#(1) 変数 st に要素 1 を追加
st.add(1) #集合は要素の重複を許さず、順序を持たないデータ構造,インデックスを指定して先頭や末尾などの要素を取得することはできません
print(st)
#(2) 変数 st から要素 1 を削除
st.remove(1)#集合型オブジェクトの remove() メソッドは指定した要素を削除
print(st)

```

```python
x={1, 2, 3}
y={3, 4, 5}
print(x.union(y))#集合型オブジェクトの union() メソッドは、指定した集合との和集合を返します。
print(x.intersection(y))#集合型オブジェクトの intersection() メソッドは、指定した集合との共通部分を返します
#以下のコードでも同じ
st = {1, 2, 3} | {3, 4, 5}
print(st)  # {1, 2, 3, 4, 5} が出力される
st = {1, 2, 3} & {3, 4, 5}
print(st)  # {3} が出力される

```

```python
def fizz_buzz(x):
    if x % 15 == 0:
        return "Fizz Buzz"
    elif x % 3 == 0:
        return "Fizz" 
    elif x % 5 == 0:
        return "Buzz"
    else:
        return x
```

```python
x1 = fizz_buzz(3)
print(x1)
x2 = fizz_buzz(5)
print(x2)
x3 = fizz_buzz(30)
print(x3)
x4 = fizz_buzz(49)
print(x4)
```

```python
class Calculator(object):
    def __init__(self, s):
        self.s = s
    def double_str(self):
        return self.s * 2 
    def triple_str(self):
        return self.s * 3
```

```python
calc = Calculator('Python')
r1 = calc.double_str()
r2 = calc.triple_str()
print(r1, r2)
```
