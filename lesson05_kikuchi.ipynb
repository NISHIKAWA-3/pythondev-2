```python
def is_kaibun(s):
    if s == s[::-1]: # 元の値とひっくり返した値が同値である
        return True
    else:

        return False

s = "にんてんどううどんてんに"
result = is_kaibun(s)
if result: # True
    print(f"「{s}」は回文です！")
else: # False
    print(f"「{s}」は回文ではないようです。")

s = "Pythonは便利だが遅い"
result = is_kaibun(s)
if result: # True
    print(f"「{s}」は回文です！")
else: # False
    print(f"「{s}」は回文ではないようです。")

s = "no devil lived on"
result = is_kaibun(s)
if result: # True
    print(f"「{s}」は回文です！")
else: # False
    print(f"「{s}」は回文ではないようです。")
```

```python
#@title 九九のリストを作成する
list = [[(a*b)for a in range(1, 10)] for b in range(1, 10)]
list
```

```python
#@title 九九の表を表示する
for v in list:
    for h in v:
        print(f"{h:3d}", end="")
    
    print("")
```
