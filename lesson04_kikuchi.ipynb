```python
count = 0
total = 0

while True:
    score = int(input("点数を入力してください:"))
    print(f"点数を入力してください:{score}")
    if score == -1:
        break

    count += 1
    total += score

print(count)
print(total)

score_average = total / count
round_score_average =round(score_average, 1)

print(f"{count}人のテストの平均点は{round_score_average}点です")
```

```python
try:
    score = int(input("点数を入力してください: "))
except ValueError:
    print("エラー: 数値を入力してください")
except:
    print("エラー: 何らかのエラーが発生しました")
else:
    if score >= 60:
        print("合格です")
        print("おめでとうございます！")
finally:
    print("処理を終了します")
```

```python
try:
    a = int(input("割られる数を入力してください："))
    print(f"割られる数を入力してください：{a}")
    b = int(input("割る数を入力してください："))
    print(f"割る数を入力してください：{b}")
    c = a / b
    print(f"{a} ÷ {b} = {c}")
except ValueError:
    print("数値を入力してください")   
except ZeroDivisionError:
    print("エラー: 0で割り算しないでください")
    
print("処理を終了します")
```
