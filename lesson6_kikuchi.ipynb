```python
# リスト
list_example = ['apple', 'banana', 'cherry']
list_example.append('date')  # リストに要素を追加
print(list_example)  # Output: ['apple', 'banana', 'cherry', 'date']
# タプル
tuple_example = ('apple', 'banana', 'cherry')
# tuple_example.append('date')  # エラー、タプルは変更不可能
print(tuple_example)  # Output: ('apple', 'banana', 'cherry')
# 辞書
dict_example = {'name': 'John', 'age': 30, 'city': 'New York'}
dict_example['city'] = 'Los Angeles'  # 辞書の要素を変更
print(dict_example)  # Output: {'name': 'John', 'age': 30, 'city': 'Los Angeles'}
# 集合
set_example = {'apple', 'banana', 'cherry'}
set_example.add('date')  # 集合に要素を追加
print(set_example)  # Output: {'date', 'cherry', 'banana', 'apple'}
# 順序は保持されない、また出力時の順序はプログラム実行ごとに異なる可能性がある
```

```python
# dict
pref_code = {'Hokkaido': 1, 'Aomori': 2, 'Tokyo': 13, 'Osaka': 27}

# keys
for code in pref_code.keys():
    print(code)

# values
for value in pref_code.values():
    print(value)

# items
for key, value in pref_code.items():
    print(f"{key} : {value}")
    
```

```python
user_inputs = []

# ユーザーからの入力を受け付け、リストに追加
print("何か入力してください（終了するには'q'を入力）:")

while True:
    user_input = input()
    if user_input == 'q':
        break # 'q'が入力されたらループを終了
    user_inputs.append(user_input)
    print(f"'{user_input}'をリストに追加しました。現在のリスト: {user_inputs}")

print("\n全ての入力:")
print(user_inputs)
```
